```python
## if there are null values in data columns ,then we can transform that data with there median by grouping categorically
#eg. df['Review Rating'] = df.groupby('Category').['Review Rating'].transform(lambda x : x.fillna(median()))
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
```

```python
df.head()
```

```python
df[['age','age_group']].head(10)
```

```python
# create purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
import pyodbc
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

connection_string = (
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=PRINCE;"
    "DATABASE=Customer_Behaviour;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

# Test the connection first
try:
    conn = pyodbc.connect(connection_string)
    print("SQL Server connected successfully!")

except Exception as e:
    print("Connection error:", e)
```

```python
engine = create_engine(
    "mssql+pyodbc:///?odbc_connect="
    + quote_plus(connection_string)
)

print("Engine created successfully!")
```

```python
with engine.connect() as connection:
    result = pd.read_sql(
        "SELECT DB_NAME() AS DatabaseName",
        connection
    )

display(result)
```

```python
df.to_sql(
    name="customer_data",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False,
    chunksize=1000
)

print("Dataset uploaded successfully!")
```
